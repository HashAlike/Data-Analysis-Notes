>Take a first look at the data

```python
import pandas as pd
import numpy as np 

flights_data= pd.read_csv(r"C:\Users\musta\Downloads\2008.csv")

# set seed for reproducibility
np.random.seed(0) 
```

```python
flights_data.head()
```

>We can see some NaN values.

```python
missing_values_count= flights_data.isnull().sum() # -> With this code, we can see how many NaN values there are in each row.

missing_values_count[0:10] # -> To find the NaN values in the first ten columns, this line of code can be used.
```

```python
#   First, we need to find the total number of missing values.
total_cells =  np.prod(flights_data.shape)
total_missing = missing_values_count.sum()
```

```python
#   Now we need to find out what percentage of the data is missing.

percent_missing= (total_missing/ total_cells)*100 
print(f"The missing data is equal to {percent_missing} of all data.")
```

>The most fundamental question we can use to determine why a value is missing is:

>“Is this value missing because it wasn't recorded or because it doesn't exist?”

>If a value is missing because it doesn't exist, it doesn't make sense to try to guess what that value might be. You'll probably want to keep these values as NaN.

>On the other hand, if a value is missing because it was not recorded, you can try to estimate what it might be based on other values in that column and row. This is called imputation.

```python
# look at the number of missing points in the first fifteen columns
missing_values_count[0:15]
```

>Drop missing values


```python
# remove all the rows that contain a missing value
flights_data.dropna()
```

>All our data appears to be missing. The reason is simple: every row contains at least one missing value.

```python
#    To fix this, it makes more sense to delete all columns containing at least one missing value rather than every row containing at least one missing value.

columns_with_na_dropped= flights_data.dropna(axis=1)
columns_with_na_dropped.head()
```

```python
# just how much data did we lose?
print("Columns in original dataset: %d \n" % flights_data.shape[1])
print("Columns with na's dropped: %d" % columns_with_na_dropped.shape[1])
```

>It appears that we have missing values in 16 columns.

>Filling in missing values automatically


```python
flights_data.columns
```

```python
#   get a small subset of the Flights dataset

subset_flights = flights_data.loc[:, [
    'DepDelay',       # numeric
    'ArrDelay',       # numeric
    'AirTime',        # numeric
    'TaxiIn',         # numeric
    'Year',           # categorical-like
    'Cancelled',      # logical
    'CancellationCode' # categorical
]]


subset_flights.head()
```

```python
#   replace all NA's with 0

subset_flights.fillna(0)
```

>Another method is to replace missing values with the value in the next column.
(This may be useful in data where the observations have a logical sequence)

```python
#   replace all NA's the value that comes directly after it in the same column, 
#   then replace all the remaining na's with 0


subset_flights.fillna(method='bfill', axis=0).fillna(0)
```
